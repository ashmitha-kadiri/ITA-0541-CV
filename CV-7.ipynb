```
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('THE-THREE.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

big = cv2.resize(img_rgb,None,fx=2,fy=2)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(big)
plt.title("Scaled Image")

plt.show()
```
