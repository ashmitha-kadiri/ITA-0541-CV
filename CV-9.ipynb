```
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('TANJIRO.jpg',0)

sobel = cv2.Sobel(img,cv2.CV_64F,1,1,ksize=5)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(sobel,cmap='gray')
plt.title("Sobel")

plt.show()
```
