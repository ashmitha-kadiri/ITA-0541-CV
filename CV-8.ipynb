```
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('TANJIRO.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rot = cv2.rotate(img_rgb,cv2.ROTATE_90_CLOCKWISE)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(rot)
plt.title("90 Degree Rotation")

plt.show()
```
