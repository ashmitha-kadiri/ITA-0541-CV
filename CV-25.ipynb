```
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
file = list(files.upload().keys())[0]

# Read image
img = cv2.imread(file)

# Convert RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Copy image
output = img_rgb.copy()

# Add label
cv2.putText(output, 'Watch Detected',
            (50,50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1, (255,0,0), 2)

# Display images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Recognition Output")
plt.imshow(output)
plt.axis('off')

plt.show()
```
