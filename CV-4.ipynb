```
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('ZENITSU.jpg',0)

equal = cv2.equalizeHist(img)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(equal,cmap='gray')
plt.title("Histogram Equalized")

plt.show()
```
